## 1. Tải dữ liệu

```python
import pandas as pd
df_train = pd.read_csv('../data/01_raw/train.csv')
df_test = pd.read_csv('../data/01_raw/test.csv')
```

## 2. Kiểm tra dữ liệu từ train

```python
df_train.head()
df_train.info()
# Tạo DataFrame mô tả
df_train.describe()
```

## Xử lí giá trị thiếu

## Cột có nhiều giá trị thiếu nhưng ý nghĩa là không có

Ví dụ các cột như: Alley, FireplaceQu, PoolQC, Fence, MiscFeature, GarageType, GarageFinish, GarageQual, GarageCond, BsmtQual, BsmtCond, BsmtExposure, BsmtFinType1, BsmtFinType2, MasVnrType

```python
cols_none = [
    'Alley', 'FireplaceQu', 'PoolQC', 'Fence', 'MiscFeature',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2', 'MasVnrType'
]
df_train[cols_none] = df_train[cols_none].fillna('None')
df_test[cols_none] = df_test[cols_none].fillna('None')
```

```python
df_train.head(10)
```

## Các cột số bị thiếu như MasVnrArea(Diện tích tường ốp gạch)

Ngta bỏ trống chắc điền 0 vào sẽ hợp lí hơn

```python
df_train['MasVnrArea'] = df_train['MasVnrArea'].fillna(0)
df_train['GarageYrBlt'] = df_train['GarageYrBlt'].fillna(0)

df_test['MasVnrArea'] = df_test['MasVnrArea'].fillna(0)
df_test['GarageYrBlt'] = df_test['GarageYrBlt'].fillna(0)
```

```python
df_train['MasVnrArea'].isnull().sum()
df_train['MasVnrArea'].head()
```

## Điền giá trị phổ biến cho các cột thiếu ít

```python
df_train['Electrical'] = df_train['Electrical'].fillna(df_train['Electrical'].mode()[0])
df_test['Electrical'] = df_test['Electrical'].fillna(df_test['Electrical'].mode()[0])
```

## Kiểm tra lại thử

```python
print(df_train.isnull().sum().sum())
print(df_test.isnull().sum().sum())
```

```python
df_train.info()
```

Vừa khớp 259 giá trị thiếu ở LotFrontage 

## Điền giá trị LotFrontage bị thiếu bằng median theo từng Neighborhood

```python
df_train['LotFrontage'] = df_train.groupby('Neighborhood')['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)
df_test['LotFrontage'] = df_test.groupby('Neighborhood')['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)
```
